```python
import pandas as pd

with open("../data/secchi_match_Zsd_0D_9-9Valid_SatRangeLessThan3_RatioBelow2_Secchi_Zsd_LinLin_selected.csv") as f:
    header = f.readline().strip().split(",")
    rows = [l.strip().split(",") for l in f.readlines()]

df = pd.DataFrame([r[:47] for r in rows], columns=header)
df.to_parquet("../data/Parquet/satellite.parquet", index=False)
print(f"Saved satellite.parquet — {len(df)} rows, {len(df.columns)} columns")
```

```python
import pandas as pd

secchi = pd.read_parquet("../data/Parquet/New_estchlsummed.parquet")
satellite = pd.read_parquet("../data/Parquet/satellite.parquet")

secchi_ids = set(secchi["Cast_ID"].dropna())
satellite_ids = set(satellite["Cast_ID"].dropna())

matching = secchi_ids & satellite_ids

print(f"secchi    — unique Cast_IDs : {len(secchi_ids)}")
print(f"satellite — unique Cast_IDs : {len(satellite_ids)}")
print(f"Exact matches               : {len(matching)}")

if matching:
    print(f"\nSample matching IDs:")
    for i in list(matching)[:5]:
        print(f"  {i!r}")
```

There are 2 Cast ID columns in Secchi (New_estchlsummed.parquet): "CAST_ID" from the CTD data and "Cast_ID" from the original Secchi data (since this dataset was initially 2 separate datasets merged together).

So, by using the "Cast_ID" column in Secchi to merge with the satellite dataset, the names line up and there are 274 matches.
