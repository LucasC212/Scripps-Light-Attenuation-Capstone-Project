```python
import pandas as pd

with open("../data/secchi_match_Zsd_0D_9-9Valid_SatRangeLessThan3_RatioBelow2_Secchi_Zsd_LinLin_selected.csv") as f:
    header = f.readline().strip().split(",")
    rows = [l.strip().split(",") for l in f.readlines()]

df = pd.DataFrame([r[:47] for r in rows], columns=header)
df.to_parquet("../data/Parquet/satellite.parquet", index=False)
print(f"Saved satellite.parquet — {len(df)} rows, {len(df.columns)} columns")
```
